```
############################################################
# SANITY CHECK: Base vs Fine-tuned E5 Embedding Model
############################################################
import torch
import numpy as np
from sentence_transformers import SentenceTransformer
import pyarrow.parquet as pq
import time

DRIVE = '/content/drive/MyDrive/hukuk-rag'
CKPT  = f'{DRIVE}/models/e5-checkpoints/checkpoint-10000'
BASE  = 'intfloat/multilingual-e5-large'

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")

# ── Load base model ──
print("\nLoading BASE model...")
t0 = time.time()
base_model = SentenceTransformer(BASE, device=str(device))
print(f"  Loaded in {time.time()-t0:.1f}s")

# ── Load fine-tuned model ──
print("\nLoading FINE-TUNED model (checkpoint-10000)...")
t0 = time.time()
tuned_model = SentenceTransformer(CKPT, device=str(device))
print(f"  Loaded in {time.time()-t0:.1f}s")
```

```
############################################################
# TEST 1: Semantic Gap on Held-out Triplets
# Use a random sample from triplets.parquet (training data)
# AND from the QA datasets (unseen during training)
############################################################
import pandas as pd

# ── Helper: encode with e5 prefix ──
def encode(model, texts, prefix="query: ", batch_size=32):
    prefixed = [f"{prefix}{t}" for t in texts]
    return model.encode(prefixed, normalize_embeddings=True,
                        show_progress_bar=False, batch_size=batch_size)

def cosine_sim(a, b):
    return np.dot(a, b)  # already normalized

# ── Sample from triplets (in-distribution) ──
triplets = pq.read_table(f'{DRIVE}/data/processed/triplets.parquet').to_pandas()
np.random.seed(42)
sample_idx = np.random.choice(len(triplets), size=50, replace=False)
trip_sample = triplets.iloc[sample_idx].reset_index(drop=True)

print(f"Sampled {len(trip_sample)} triplets for evaluation")
print(f"Total triplets: {len(triplets)}")

# ── Sample from QA dataset (out-of-distribution for training) ──
qa = pd.read_parquet(f'{DRIVE}/data/raw/turkish_law_qa.parquet')
qa_sample = qa.sample(50, random_state=42).reset_index(drop=True)
print(f"Sampled {len(qa_sample)} QA pairs for OOD evaluation")
```

```
############################################################
# TEST 1A: Triplet Gap Test (in-distribution)
# For each triplet: sim(query, positive) - sim(query, negative)
# Fine-tuned model should have LARGER gap
############################################################

queries = trip_sample['query'].tolist()
positives = trip_sample['positive'].tolist()
negatives = trip_sample['negative'].tolist()

# Base model
print("Encoding with BASE model...")
b_q = encode(base_model, queries, prefix="query: ")
b_p = encode(base_model, positives, prefix="passage: ")
b_n = encode(base_model, negatives, prefix="passage: ")

# Fine-tuned model
print("Encoding with TUNED model...")
t_q = encode(tuned_model, queries, prefix="query: ")
t_p = encode(tuned_model, positives, prefix="passage: ")
t_n = encode(tuned_model, negatives, prefix="passage: ")

# Compute gaps
base_gaps = []
tuned_gaps = []
for i in range(len(queries)):
    bg = cosine_sim(b_q[i], b_p[i]) - cosine_sim(b_q[i], b_n[i])
    tg = cosine_sim(t_q[i], t_p[i]) - cosine_sim(t_q[i], t_n[i])
    base_gaps.append(bg)
    tuned_gaps.append(tg)

base_gaps = np.array(base_gaps)
tuned_gaps = np.array(tuned_gaps)

print(f"\n{'='*60}")
print(f"TEST 1A: TRIPLET GAP (in-distribution, n={len(queries)})")
print(f"{'='*60}")
print(f"  BASE  — mean gap: {base_gaps.mean():.4f} ± {base_gaps.std():.4f}")
print(f"  TUNED — mean gap: {tuned_gaps.mean():.4f} ± {tuned_gaps.std():.4f}")
print(f"  Improvement:      {(tuned_gaps.mean() - base_gaps.mean()):.4f}")
print(f"  Tuned wins: {(tuned_gaps > base_gaps).sum()}/{len(queries)} ({100*(tuned_gaps > base_gaps).mean():.1f}%)")
print(f"  Ties (within 0.01): {(np.abs(tuned_gaps - base_gaps) < 0.01).sum()}")

# Also show absolute similarities
base_pos_sim = np.array([cosine_sim(b_q[i], b_p[i]) for i in range(len(queries))])
tuned_pos_sim = np.array([cosine_sim(t_q[i], t_p[i]) for i in range(len(queries))])
base_neg_sim = np.array([cosine_sim(b_q[i], b_n[i]) for i in range(len(queries))])
tuned_neg_sim = np.array([cosine_sim(t_q[i], t_n[i]) for i in range(len(queries))])

print(f"\n  Positive similarities:")
print(f"    BASE:  {base_pos_sim.mean():.4f} ± {base_pos_sim.std():.4f}")
print(f"    TUNED: {tuned_pos_sim.mean():.4f} ± {tuned_pos_sim.std():.4f}")
print(f"  Negative similarities:")
print(f"    BASE:  {base_neg_sim.mean():.4f} ± {base_neg_sim.std():.4f}")
print(f"    TUNED: {tuned_neg_sim.mean():.4f} ± {tuned_neg_sim.std():.4f}")
```

```
############################################################
# TEST 1B: QA Similarity Test (out-of-distribution)
# Use QA pairs where answer = "positive", random answer = "negative"
############################################################

qa_questions = qa_sample['question'].tolist()
qa_answers = qa_sample['answer'].tolist()

# Create "negatives" by shuffling answers
np.random.seed(123)
shuffled_idx = np.random.permutation(len(qa_answers))
# Make sure no answer maps to its own question
while any(shuffled_idx[i] == i for i in range(len(shuffled_idx))):
    shuffled_idx = np.random.permutation(len(qa_answers))
qa_negatives = [qa_answers[i] for i in shuffled_idx]

# Base model
print("Encoding QA with BASE model...")
b_qq = encode(base_model, qa_questions, prefix="query: ")
b_qa = encode(base_model, qa_answers, prefix="passage: ")
b_qn = encode(base_model, qa_negatives, prefix="passage: ")

# Tuned model
print("Encoding QA with TUNED model...")
t_qq = encode(tuned_model, qa_questions, prefix="query: ")
t_qa = encode(tuned_model, qa_answers, prefix="passage: ")
t_qn = encode(tuned_model, qa_negatives, prefix="passage: ")

# Compute gaps
base_qa_gaps = np.array([cosine_sim(b_qq[i], b_qa[i]) - cosine_sim(b_qq[i], b_qn[i]) for i in range(len(qa_questions))])
tuned_qa_gaps = np.array([cosine_sim(t_qq[i], t_qa[i]) - cosine_sim(t_qq[i], t_qn[i]) for i in range(len(qa_questions))])

print(f"\n{'='*60}")
print(f"TEST 1B: QA GAP (out-of-distribution, n={len(qa_questions)})")
print(f"{'='*60}")
print(f"  BASE  — mean gap: {base_qa_gaps.mean():.4f} ± {base_qa_gaps.std():.4f}")
print(f"  TUNED — mean gap: {tuned_qa_gaps.mean():.4f} ± {tuned_qa_gaps.std():.4f}")
print(f"  Improvement:      {(tuned_qa_gaps.mean() - base_qa_gaps.mean()):.4f}")
print(f"  Tuned wins: {(tuned_qa_gaps > base_qa_gaps).sum()}/{len(qa_questions)} ({100*(tuned_qa_gaps > base_qa_gaps).mean():.1f}%)")

# Absolute sims
b_pos = np.array([cosine_sim(b_qq[i], b_qa[i]) for i in range(len(qa_questions))])
t_pos = np.array([cosine_sim(t_qq[i], t_qa[i]) for i in range(len(qa_questions))])
b_neg = np.array([cosine_sim(b_qq[i], b_qn[i]) for i in range(len(qa_questions))])
t_neg = np.array([cosine_sim(t_qq[i], t_qn[i]) for i in range(len(qa_questions))])

print(f"\n  Positive similarities:")
print(f"    BASE:  {b_pos.mean():.4f} ± {b_pos.std():.4f}")
print(f"    TUNED: {t_pos.mean():.4f} ± {t_pos.std():.4f}")
print(f"  Negative similarities:")
print(f"    BASE:  {b_neg.mean():.4f} ± {b_neg.std():.4f}")
print(f"    TUNED: {t_neg.mean():.4f} ± {t_neg.std():.4f}")
```

```
############################################################
# TEST 2: Retrieval Quality on Real Chunks
# Encode a subset of chunks, build mini-FAISS, compare top-k
############################################################
import faiss

# Free one model to save VRAM — we'll reload later if needed
del base_model
torch.cuda.empty_cache()
import gc; gc.collect()
print("Freed base model from GPU")

# Load a sample of real chunks (10k is enough for a sanity check)
print("\nLoading chunk sample...")
chunks_table = pq.read_table(
    f'{DRIVE}/data/processed/chunks_filtered.parquet',
    columns=['chunk_id', 'text']
)
chunks_df = chunks_table.to_pandas()
print(f"Total chunks: {len(chunks_df)}")

# Sample 10k chunks
np.random.seed(42)
if len(chunks_df) > 10000:
    chunk_sample = chunks_df.sample(10000, random_state=42).reset_index(drop=True)
else:
    chunk_sample = chunks_df.reset_index(drop=True)
print(f"Using {len(chunk_sample)} chunks for retrieval test")

# Encode chunks with fine-tuned model
print("\nEncoding chunks with TUNED model...")
chunk_texts = chunk_sample['text'].tolist()
chunk_embs = encode(tuned_model, chunk_texts, prefix="passage: ", batch_size=64)
print(f"  Embeddings shape: {chunk_embs.shape}")

# Build FAISS index
dim = chunk_embs.shape[1]
index = faiss.IndexFlatIP(dim)  # Inner product = cosine sim (already normalized)
index.add(chunk_embs.astype(np.float32))
print(f"  FAISS index built: {index.ntotal} vectors")
```

```
############################################################
# TEST 2: Query the mini-FAISS with legal questions
############################################################

test_queries = [
    "Kasten adam öldürme suçunun cezası nedir?",
    "Kira sözleşmesinin feshi nasıl yapılır?",
    "İş kazası durumunda işverenin sorumluluğu nedir?",
    "Boşanma davasında nafaka nasıl belirlenir?",
    "Miras hukukunda saklı pay nedir?",
    "Ticaret şirketlerinin birleşme şartları nelerdir?",
    "Haksız tutuklama halinde tazminat hakkı var mıdır?",
    "Kamulaştırma bedelinin tespiti nasıl yapılır?",
    "Yargıtay kararlarına itiraz yolları nelerdir?",
    "İcra takibinde borca itiraz süresi ne kadardır?"
]

k = 5
q_embs = encode(tuned_model, test_queries, prefix="query: ")
D, I = index.search(q_embs.astype(np.float32), k)

print(f"{'='*60}")
print(f"TEST 2: RETRIEVAL ON 10K CHUNK SAMPLE (top-{k})")
print(f"{'='*60}")

for qi, query in enumerate(test_queries):
    print(f"\n{'─'*60}")
    print(f"Q{qi+1}: {query}")
    print(f"{'─'*60}")
    for rank in range(k):
        idx = I[qi][rank]
        score = D[qi][rank]
        text = chunk_sample.iloc[idx]['text'][:200]
        print(f"  [{rank+1}] score={score:.4f} | {text}...")
```

```
############################################################
# REBUILD FAISS INDEX WITH FINE-TUNED E5 (checkpoint-10000)
############################################################
# This will:
# 1. Load fine-tuned E5 model
# 2. Encode all 2.5M chunks in batches (with checkpointing)
# 3. Build IVF-PQ FAISS index
# 4. Save to Drive
############################################################

import torch, gc, os, time, json
import numpy as np
import pyarrow.parquet as pq
from pathlib import Path
from sentence_transformers import SentenceTransformer

# Paths
DRIVE = Path('/content/drive/MyDrive/hukuk-rag')
CHUNKS_PATH = DRIVE / 'data' / 'processed' / 'chunks_filtered.parquet'
NEW_FAISS_DIR = DRIVE / 'indexes' / 'finetuned'
NEW_FAISS_DIR.mkdir(parents=True, exist_ok=True)
FAISS_PATH = NEW_FAISS_DIR / 'faiss_ft.index'
MAPPING_PATH = NEW_FAISS_DIR / 'faiss_ft.mapping.pkl'
EMBEDDINGS_CACHE = NEW_FAISS_DIR / 'embeddings_cache'
EMBEDDINGS_CACHE.mkdir(parents=True, exist_ok=True)
CKPT = str(DRIVE / 'models' / 'e5-checkpoints' / 'checkpoint-10000')

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")
print(f"Chunks: {CHUNKS_PATH}")
print(f"Output: {FAISS_PATH}")

# Check total chunks
pf = pq.ParquetFile(str(CHUNKS_PATH))
total_chunks = pf.metadata.num_rows
print(f"Total chunks to encode: {total_chunks:,}")
```

```
############################################################
# STEP 1: Load fine-tuned model
############################################################
print("Loading fine-tuned E5 model...")
t0 = time.time()
model = SentenceTransformer(CKPT, device=str(device))
print(f"Loaded in {time.time()-t0:.1f}s")
print(f"GPU memory: {torch.cuda.memory_allocated()/1e9:.1f} GB")
```

```
############################################################
# STEP 2: Encode ALL chunks in batches with checkpointing
#
# Strategy: Process in mega-batches of 50k chunks.
# After each mega-batch, save embeddings to disk as .npy
# If session crashes, we resume from last saved mega-batch.
############################################################
import pickle

MEGA_BATCH = 50_000   # chunks per checkpoint file
ENCODE_BATCH = 128    # chunks per model.encode() call

def get_completed_batches():
    """Check which mega-batches are already encoded."""
    done = []
    for f in sorted(EMBEDDINGS_CACHE.glob('emb_*.npy')):
        idx = int(f.stem.split('_')[1])
        done.append(idx)
    return sorted(done)

completed = get_completed_batches()
print(f"Already completed mega-batches: {len(completed)}")
if completed:
    print(f"  Batch indices: {completed[0]}..{completed[-1]}")
    print(f"  Chunks encoded so far: ~{len(completed) * MEGA_BATCH:,}")

# Calculate which mega-batches we still need
total_mega_batches = (total_chunks + MEGA_BATCH - 1) // MEGA_BATCH
remaining = [i for i in range(total_mega_batches) if i not in completed]
print(f"Total mega-batches needed: {total_mega_batches}")
print(f"Remaining: {len(remaining)}")
```

```
############################################################
# STEP 2b: Encode chunks with resume support
############################################################
from tqdm.auto import tqdm

pf = pq.ParquetFile(str(CHUNKS_PATH))

# We'll read the parquet in row-group batches and accumulate
# until we hit MEGA_BATCH, then encode and save

chunk_buffer_texts = []
chunk_buffer_ids = []
current_mega_idx = 0
global_row = 0
start_time = time.time()

# Skip already-completed rows
skip_rows = len(completed) * MEGA_BATCH
if skip_rows > 0:
    print(f"Skipping first {skip_rows:,} rows (already encoded)")

print(f"\nEncoding {total_chunks:,} chunks in mega-batches of {MEGA_BATCH:,}...")
print(f"Checkpoint files saved to: {EMBEDDINGS_CACHE}")
print()

for batch in pf.iter_batches(batch_size=10_000, columns=['chunk_id', 'text']):
    df = batch.to_pandas()
    batch_size_actual = len(df)

    # Skip already-completed rows
    if global_row + batch_size_actual <= skip_rows:
        global_row += batch_size_actual
        continue

    # Handle partial skip at boundary
    if global_row < skip_rows:
        skip_in_batch = skip_rows - global_row
        df = df.iloc[skip_in_batch:]

    global_row += batch_size_actual

    chunk_buffer_texts.extend(df['text'].tolist())
    chunk_buffer_ids.extend(df['chunk_id'].tolist())

    # When buffer reaches mega-batch size, encode and save
    while len(chunk_buffer_texts) >= MEGA_BATCH:
        mega_texts = chunk_buffer_texts[:MEGA_BATCH]
        mega_ids = chunk_buffer_ids[:MEGA_BATCH]
        chunk_buffer_texts = chunk_buffer_texts[MEGA_BATCH:]
        chunk_buffer_ids = chunk_buffer_ids[MEGA_BATCH:]

        if current_mega_idx not in completed:
            # Add "passage: " prefix for e5
            prefixed = [f"passage: {t}" for t in mega_texts]

            # Encode in sub-batches
            embeddings = model.encode(
                prefixed,
                normalize_embeddings=True,
                show_progress_bar=False,
                batch_size=ENCODE_BATCH,
            )

            # Save to disk
            np.save(str(EMBEDDINGS_CACHE / f'emb_{current_mega_idx:04d}.npy'), embeddings)
            with open(str(EMBEDDINGS_CACHE / f'ids_{current_mega_idx:04d}.pkl'), 'wb') as f:
                pickle.dump(mega_ids, f)

            elapsed = time.time() - start_time
            done_chunks = (current_mega_idx + 1 - len(completed)) * MEGA_BATCH
            rate = done_chunks / elapsed if elapsed > 0 else 0
            eta = (total_chunks - (current_mega_idx + 1) * MEGA_BATCH) / rate if rate > 0 else 0

            print(f"  Mega-batch {current_mega_idx+1}/{total_mega_batches} saved | "
                  f"{(current_mega_idx+1)*MEGA_BATCH:,}/{total_chunks:,} chunks | "
                  f"{rate:.0f} chunks/s | ETA: {eta/60:.0f}min")

        current_mega_idx += 1

    del df; gc.collect()

# Handle remaining chunks in buffer
if chunk_buffer_texts and current_mega_idx not in completed:
    prefixed = [f"passage: {t}" for t in chunk_buffer_texts]
    embeddings = model.encode(
        prefixed,
        normalize_embeddings=True,
        show_progress_bar=False,
        batch_size=ENCODE_BATCH,
    )
    np.save(str(EMBEDDINGS_CACHE / f'emb_{current_mega_idx:04d}.npy'), embeddings)
    with open(str(EMBEDDINGS_CACHE / f'ids_{current_mega_idx:04d}.pkl'), 'wb') as f:
        pickle.dump(chunk_buffer_ids, f)
    print(f"  Mega-batch {current_mega_idx+1}/{total_mega_batches} (final, {len(chunk_buffer_texts):,} chunks) saved")

total_time = time.time() - start_time
print(f"\nEncoding complete! Total time: {total_time/60:.1f} minutes")
print(f"Checkpoint files: {len(list(EMBEDDINGS_CACHE.glob('emb_*.npy')))}")
```

```
############################################################
# CHECK: How badly are chunks being truncated by E5's tokenizer?
# E5 uses XLM-RoBERTa tokenizer with 512 max tokens.
# We chunked at 480 WORDS but the model sees SUBWORD tokens.
############################################################
from transformers import AutoTokenizer
import pyarrow.parquet as pq
import numpy as np

tokenizer = AutoTokenizer.from_pretrained("intfloat/multilingual-e5-large")
MAX_SEQ_LEN = 512
PREFIX_OVERHEAD = len(tokenizer.encode("passage: ", add_special_tokens=True))  # ~5-6 tokens

print(f"Model max sequence length: {MAX_SEQ_LEN}")
print(f"Prefix overhead ('passage: ' + special tokens): {PREFIX_OVERHEAD} tokens")
print(f"Effective limit for chunk content: {MAX_SEQ_LEN - PREFIX_OVERHEAD} tokens")

# Sample 10k chunks
pf = pq.ParquetFile('/content/drive/MyDrive/hukuk-rag/data/processed/chunks_filtered.parquet')
sample = next(pf.iter_batches(batch_size=10_000, columns=['text'])).to_pandas()

# Tokenize each chunk and measure actual subword token count
token_counts = []
for text in sample['text']:
    tokens = tokenizer.encode(f"passage: {text}", add_special_tokens=True)
    token_counts.append(len(tokens))

token_counts = np.array(token_counts)
word_counts = sample['text'].str.split().str.len().values

print(f"\n{'='*60}")
print(f"SUBWORD TOKEN ANALYSIS (n={len(sample):,} chunks)")
print(f"{'='*60}")
print(f"\nWhitespace word counts:")
print(f"  Mean: {word_counts.mean():.0f}, Median: {np.median(word_counts):.0f}")
print(f"  P90: {np.percentile(word_counts, 90):.0f}, P99: {np.percentile(word_counts, 99):.0f}, Max: {word_counts.max()}")

print(f"\nSubword token counts (what the model actually sees):")
print(f"  Mean: {token_counts.mean():.0f}, Median: {np.median(token_counts):.0f}")
print(f"  P90: {np.percentile(token_counts, 90):.0f}, P99: {np.percentile(token_counts, 99):.0f}, Max: {token_counts.max()}")

ratio = token_counts / word_counts
print(f"\nSubword/word ratio:")
print(f"  Mean: {ratio.mean():.2f}x, Median: {np.median(ratio):.2f}x")
print(f"  (i.e., each word becomes ~{ratio.mean():.1f} subword tokens on average)")

truncated = (token_counts > MAX_SEQ_LEN).sum()
pct = 100 * truncated / len(token_counts)
print(f"\n🚨 TRUNCATION:")
print(f"  Chunks exceeding {MAX_SEQ_LEN} tokens: {truncated:,}/{len(token_counts):,} ({pct:.1f}%)")

if truncated > 0:
    overflow = token_counts[token_counts > MAX_SEQ_LEN] - MAX_SEQ_LEN
    print(f"  Avg tokens lost per truncated chunk: {overflow.mean():.0f}")
    print(f"  Max tokens lost: {overflow.max()}")
    pct_lost = overflow / token_counts[token_counts > MAX_SEQ_LEN] * 100
    print(f"  Avg % of chunk content lost: {pct_lost.mean():.1f}%")
```

```
############################################################
# FIXED v2: Skip rows efficiently using row counter, not loading
############################################################
import torch, gc, os, time, pickle
import numpy as np
import pyarrow.parquet as pq
from pathlib import Path
from sentence_transformers import SentenceTransformer

DRIVE = Path('/content/drive/MyDrive/hukuk-rag')
CHUNKS_PATH = DRIVE / 'data' / 'processed' / 'chunks_filtered.parquet'
EMBEDDINGS_CACHE = DRIVE / 'indexes' / 'finetuned' / 'embeddings_cache'
CKPT = str(DRIVE / 'models' / 'e5-checkpoints' / 'checkpoint-10000')

MEGA_BATCH = 50_000
ENCODE_BATCH = 128

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
pf = pq.ParquetFile(str(CHUNKS_PATH))
total_chunks = pf.metadata.num_rows
total_mega_batches = (total_chunks + MEGA_BATCH - 1) // MEGA_BATCH

# Check completed
completed = sorted([int(f.stem.split('_')[1]) for f in EMBEDDINGS_CACHE.glob('emb_*.npy')])
skip_rows = (max(completed) + 1) * MEGA_BATCH if completed else 0

print(f"Device: {device}")
print(f"Total chunks: {total_chunks:,}")
print(f"Completed batches: {len(completed)} (0..{max(completed) if completed else 'none'})")
print(f"Skipping first {skip_rows:,} rows")
print(f"Remaining: {total_chunks - skip_rows:,} chunks")

# Load model
print("\nLoading model...")
t0 = time.time()
model = SentenceTransformer(CKPT, device=str(device))
print(f"Loaded in {time.time()-t0:.1f}s")

# Encode remaining chunks — skip rows by counter, not by loading
chunk_buffer_texts = []
chunk_buffer_ids = []
current_mega_idx = max(completed) + 1 if completed else 0
row_counter = 0
start_time = time.time()

print(f"\nStarting from mega-batch {current_mega_idx+1}/{total_mega_batches}...")

for batch in pf.iter_batches(batch_size=10_000, columns=['chunk_id', 'text']):
    batch_len = batch.num_rows

    # Skip rows we've already encoded — don't convert to pandas
    if row_counter + batch_len <= skip_rows:
        row_counter += batch_len
        continue

    df = batch.to_pandas()

    # Partial skip at boundary
    if row_counter < skip_rows:
        df = df.iloc[skip_rows - row_counter:]

    row_counter += batch_len

    chunk_buffer_texts.extend(df['text'].tolist())
    chunk_buffer_ids.extend(df['chunk_id'].tolist())

    while len(chunk_buffer_texts) >= MEGA_BATCH:
        mega_texts = chunk_buffer_texts[:MEGA_BATCH]
        mega_ids = chunk_buffer_ids[:MEGA_BATCH]
        chunk_buffer_texts = chunk_buffer_texts[MEGA_BATCH:]
        chunk_buffer_ids = chunk_buffer_ids[MEGA_BATCH:]

        prefixed = [f"passage: {t}" for t in mega_texts]
        embeddings = model.encode(
            prefixed,
            normalize_embeddings=True,
            show_progress_bar=False,
            batch_size=ENCODE_BATCH,
        )
        np.save(str(EMBEDDINGS_CACHE / f'emb_{current_mega_idx:04d}.npy'), embeddings)
        with open(str(EMBEDDINGS_CACHE / f'ids_{current_mega_idx:04d}.pkl'), 'wb') as f:
            pickle.dump(mega_ids, f)

        elapsed = time.time() - start_time
        new_done = (current_mega_idx - (max(completed) if completed else -1)) * MEGA_BATCH
        rate = new_done / elapsed if elapsed > 0 else 0
        remaining = total_chunks - (current_mega_idx + 1) * MEGA_BATCH
        eta = remaining / rate if rate > 0 else 0

        print(f"  Mega-batch {current_mega_idx+1}/{total_mega_batches} saved | "
              f"{rate:.0f} chunks/s | ETA: {eta/60:.0f}min")

        current_mega_idx += 1

    del df
    gc.collect()

# Final partial batch
if chunk_buffer_texts:
    prefixed = [f"passage: {t}" for t in chunk_buffer_texts]
    embeddings = model.encode(
        prefixed, normalize_embeddings=True,
        show_progress_bar=False, batch_size=ENCODE_BATCH,
    )
    np.save(str(EMBEDDINGS_CACHE / f'emb_{current_mega_idx:04d}.npy'), embeddings)
    with open(str(EMBEDDINGS_CACHE / f'ids_{current_mega_idx:04d}.pkl'), 'wb') as f:
        pickle.dump(chunk_buffer_ids, f)
    print(f"  Mega-batch {current_mega_idx+1}/{total_mega_batches} (final, {len(chunk_buffer_texts):,} chunks) saved")

total_time = time.time() - start_time
print(f"\nEncoding complete! Total time: {total_time/60:.1f} minutes")
print(f"Checkpoint files: {len(list(EMBEDDINGS_CACHE.glob('emb_*.npy')))}")
```

```
############################################################
# STEP 3: Build FAISS IVF-PQ index from cached embeddings
############################################################
import numpy as np
import pickle
import faiss
import time
from pathlib import Path

DRIVE = Path('/content/drive/MyDrive/hukuk-rag')
CACHE = DRIVE / 'indexes' / 'finetuned' / 'embeddings_cache'
FAISS_PATH = DRIVE / 'indexes' / 'finetuned' / 'faiss_ft.index'
MAPPING_PATH = DRIVE / 'indexes' / 'finetuned' / 'faiss_ft.mapping.pkl'

# Load all embeddings and IDs
print("Loading cached embeddings...")
t0 = time.time()
all_embs = []
all_ids = []

for i in range(50):
    emb_path = CACHE / f'emb_{i:04d}.npy'
    ids_path = CACHE / f'ids_{i:04d}.pkl'
    emb = np.load(str(emb_path))
    with open(str(ids_path), 'rb') as f:
        ids = pickle.load(f)
    all_embs.append(emb)
    all_ids.extend(ids)
    if (i+1) % 10 == 0:
        print(f"  Loaded {i+1}/50 files")

embeddings = np.vstack(all_embs).astype(np.float32)
del all_embs
print(f"  Total embeddings: {embeddings.shape}")
print(f"  Total IDs: {len(all_ids)}")
print(f"  Loaded in {time.time()-t0:.1f}s")

# Build IVF-PQ index
dim = embeddings.shape[1]
nlist = 256
m = 32
nbits = 8

print(f"\nBuilding FAISS IVF-PQ index (dim={dim}, nlist={nlist}, m={m})...")
quantizer = faiss.IndexFlatIP(dim)
index = faiss.IndexIVFPQ(quantizer, dim, nlist, m, nbits)

# Train on sample
np.random.seed(42)
train_size = min(len(embeddings), nlist * 40)
train_idx = np.random.choice(len(embeddings), train_size, replace=False)
print(f"  Training on {train_size:,} vectors...")
index.train(embeddings[train_idx])

# Add all vectors
print(f"  Adding {len(embeddings):,} vectors...")
index.add(embeddings)
print(f"  Index built: {index.ntotal:,} vectors")

# Save index
faiss.write_index(index, str(FAISS_PATH))
print(f"  Saved index to {FAISS_PATH}")

# Build and save mapping (just chunk_id for now)
# We need to load chunk text for the mapping
import pyarrow.parquet as pq
print("\nBuilding chunk mapping...")
pf = pq.ParquetFile(str(DRIVE / 'data' / 'processed' / 'chunks_filtered.parquet'))
chunk_texts = []
chunk_meta = []
for batch in pf.iter_batches(batch_size=50_000):
    df = batch.to_pandas()
    for _, row in df.iterrows():
        chunk_texts.append(row['text'])
        chunk_meta.append({k: row[k] for k in df.columns if k != 'text'})

mapping = []
for i, cid in enumerate(all_ids):
    # Find matching text by chunk_id
    mapping.append({"chunk_id": cid, "text": chunk_texts[i] if i < len(chunk_texts) else ""})

with open(str(MAPPING_PATH), 'wb') as f:
    pickle.dump(mapping, f)
print(f"  Saved mapping ({len(mapping):,} entries) to {MAPPING_PATH}")

del embeddings
import gc; gc.collect()
print(f"\nDone! Total time: {(time.time()-t0)/60:.1f} minutes")
```
